create a class circle with a constructor initializing the radius , another fuction to calculate the area of the circle, another function circumference of the circle.

```python
class circle:
  def __init__(self,radius):
    self.radius=radius
  def area(self):
    print(f"Area of the circle is: {3.14*self.radius*self.radius}")
  def circumference(self):
    print(f"Circumference of the circle is: {2*3.14*self.radius}")
c1=circle(5)
c1.area()
c1.circumference()
```

there are two classes naming num1 and num2, in num1 class there are two variables a and b that need to be initialized uusing constructor. it also have a memeber function show value to print this values.num2 class is inherited from num1 and it has two function 1 one is show some that show the sum of the variable the second one is to show product that display the product of the variable.

```python
class num1:
  def __init__(self,a,b):
    self.a=a
    self.b=b
  def show_value(self):
    print(f"Value of a is: {self.a}")
    print(f"Value of the b: {self.b}")
class num2(num1):
  def show_sum(self):
    print(self.a+self.b)
  def show_product(self):
    print(self.a*self.b)
n1=num2(2,3)
n1.show_value()
n1.show_sum()
n1.show_product()

```

there are two classes student and master ,student class is the base class and the master class is the derived class the class student has two memeber variable id and mark.there is constructor to initialize this variables and a function show result which will check the mark for pass and fail.if the mark is grater than or equal to 40 it is a pass otherwise it is fail.similarly in the masters student contain a function show result that will check whether the mark is greater than or equal to 50.if thats the case then it is pass otherwise it will be a fail.initialize the student object and a master student object and call the function respectively.

```python
class student:
  def __init__(self,id,mark):
    self.id=id
    self.mark=mark
  def show_result(self):
    if(self.mark>=40):
      print(f"the student with id {self.id} is pass")
    else:
      print(f"the student with id {self.id} is fail")
class master(student):
  def show_mark(self):
    if(self.mark>=50):
      print(f"the student with id {self.id} is pass")
    else:
      print(f"the student with id {self.id} is fail")
s1=master(101,60)
s1.show_result()
m1=master(102,30)
m1.show_mark()
```
